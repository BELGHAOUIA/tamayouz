````python
import base64
import json
import os
from decouple import config
from PIL import Image
from google import genai


# Access the API key
gemini_api_key = config("GEMINI_API_KEY")

# 1. Initialize the client
client = genai.Client(api_key=gemini_api_key)

try:
    # 2. Open the image you want to analyze
    image_path = "bulletin.jpg"  # Replace with your actual filename
    img = Image.open(image_path)
   
    prompt=""
    with open("prompt.txt", "r", encoding="utf-8") as f:
        prompt = f.read()

    # 3. Call the API using a multimodal model
    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=[img, prompt],
    )

    # 4. Parse the response text as JSON
    raw_content = response.text.strip()

    # Strip accidental code fences if the model adds them
    if raw_content.startswith("```"):
        raw_content = raw_content.strip("`")
        raw_content = raw_content.replace("json\n", "", 1).strip()

    try:
        result = json.loads(raw_content)
    except json.JSONDecodeError:
        result = {"error": "Failed to parse JSON", "raw_response": raw_content}

    print("\n--- Gemini Analysis ---")

    # 5. Optional: save to file
    with open("extracted_text.json", "w") as f:
        json.dump(result, f, indent=2)

except FileNotFoundError:
    print(f"Error: The file '{image_path}' could not be found.")
except Exception as e:
    print(f"An API error occurred: {e}")
````
